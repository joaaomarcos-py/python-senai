```python
livros = {}
usuarios = {}

def cadastrar_livros():
    idlivro = input("ID do livro: ")
    titulo = input("Título do livro: ")

    livros[idlivro] = {
        "titulo": titulo,
        "disponivel": True
    }

    print("Livro cadastrado!\n")


def listar_livros():
    if not livros:
        print("Nenhum livro cadastrado.\n")
        return

    for idlivro, dados in livros.items():
        status = "Disponível" if dados["disponivel"] else "Indisponível"
        print(f"ID: {idlivro} | Título: {dados['titulo']} | Status: {status}")
    print()


def cadastrar_usuario():
    idusuario = input("ID do usuário: ")
    nome = input("Nome do usuário: ")

    usuarios[idusuario] = nome
    print("Usuário cadastrado!\n")


def emprestimo():
    idlivro = input("ID do livro: ")
    idusuario = input("ID do usuário: ")

    if idlivro not in livros:
        print("Livro não encontrado.\n")
        return

    if idusuario not in usuarios:
        print("Usuário não encontrado.\n")
        return

    if livros[idlivro]["disponivel"]:
        livros[idlivro]["disponivel"] = False
        print("Empréstimo realizado.\n")
    else:
        print("Livro indisponível.\n")


def devolver_livro():
    idlivro = input("ID do livro: ")

    if idlivro not in livros:
        print("Livro não encontrado.\n")
        return

    if not livros[idlivro]["disponivel"]:
        livros[idlivro]["disponivel"] = True
        print("Livro devolvido.\n")
    else:
        print("Esse livro já está disponível.\n")

def menu():
    while True:
        print("MENU")
        print("1 - Cadastrar livro")
        print("2 - Listar livros")
        print("3 - Cadastrar usuário")
        print("4 - Empréstimo")
        print("5 - Devolver livro")
        print("0 - Sair")

        opcao = input("Escolha: ")

        if opcao == "1":
            cadastrar_livros()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            cadastrar_usuario()
        elif opcao == "4":
            emprestimo()
        elif opcao == "5":
            devolver_livro()
        elif opcao == "0":
            print("Saindo...")
            break
        else:
            print("Opção inválida.\n")


menu()

```
