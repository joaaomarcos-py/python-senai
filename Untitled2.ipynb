```python
def verifica_numero(n):
    if n > 0 :
        return "P"
    else:
        return "N"
numero_digitado = int(input("Digite um número: "))
resultado = verifica_numero(numero_digitado)

print("Resultado:", resultado)
```

```python
def conta_char(P):
  if len(P) > 0:
    print(f"Sua palavra tem {len(P)} caracteres")
  else:
    print("Passar um char positivo")

conta_char("")
```
